Дана история поисковых запросов: каждая строка — один введённый запрос. Повтор в списке значит, что этот запрос ввели ещё раз.

```python
queries = [
    "чехол",
    "iphone",
    "чехол",
    "наушники",
    "iphone",
    "iphone",
    "кабель",
    "чехол",
    "iphone",
]
```

Нужно найти:

- сколько всего поисковых запросов в ленте
- сколько раз ввели каждый запрос
- какой запрос вводили чаще всего
- какую долю всех поисков он занимает
- какие запросы встретились один раз

````python
from collections import Counter
queries = [
    "чехол",
    "iphone",
    "чехол",
    "наушники",
    "iphone",
    "iphone",
    "кабель",
    "чехол",
    "iphone",
]

print(f'''Сколько всего поисковых запросов в ленте - {len(queries)}
Сколько раз ввели каждый запрос - {Counter(queries)}
Какой запрос вводили чаще всего - {Counter(queries).most_common(1)[0][0]}
Какую долю всех поисков он занимает - {Counter(queries).most_common(1)[0][1] / len(queries)}''')
list_name = []
for k, v in Counter(queries).items():
    if v == 1:
        list_name.append(k)
print(f'Kакие запросы встретились один раз - {list_name}')
````

Есть список заказов:

status - статус заказа

amount — сумма заказа.
orders = [

    {"id": 3, "buyer": "anya", "status": "delivered", "amount": 1_500},
    {"id": 4, "buyer": "vera", "status": "delivered", "amount": 3_200},
    {"id": 5, "buyer": "boris", "status": "delivered", "amount": 700},

Нужно найти:

на какую сумму оформили возвраты
кто хотя бы раз вернул заказ
сколько заказов доставлено покупателю
сколько заказов доставлено покупателю

````python
orders = [
    {"id": 1, "buyer": "anya", "status": "delivered", "amount": 900},
    {"id": 2, "buyer": "boris", "status": "returned", "amount": 4_500},
    {"id": 3, "buyer": "anya", "status": "delivered", "amount": 1_500},
    {"id": 4, "buyer": "vera", "status": "delivered", "amount": 3_200},
    {"id": 5, "buyer": "boris", "status": "delivered", "amount": 700},
    {"id": 6, "buyer": "gleb", "status": "returned", "amount": 2_100},
]
sum_returned = 0
people_returned = set()
sum_delivered = []
for i in orders:
    if i['status'] == 'returned':
        sum_returned += i['amount']
        people_returned.add(i['buyer'])
    else:
        sum_delivered.append(i['amount'])
print(f'''На какую сумму оформили возвраты - {sum_returned}
Кто хотя бы раз вернул заказ - {people_returned}
Сколько заказов доставлено покупателю - {len(sum_delivered)}
Сколько заказов доставлено покупателю - {sum(sum_delivered) / len(sum_delivered)}''')
````

Есть статистика за пять дней работы магазина. У каждого дня число заказов, выручка и число возвратов.

days = [
    {"day": "пн", "orders": 20, "revenue": 40_000, "returns": 2},
    {"day": "вт", "orders": 16, "revenue": 19_200, "returns": 4},
    {"day": "ср", "orders": 25, "revenue": 55_000, "returns": 1},
    {"day": "чт", "orders": 10, "revenue": 12_000, "returns": 3},
    {"day": "пт", "orders": 30, "revenue": 48_000, "returns": 3},
]
Нужно найти:

выручку за всю неделю
день с самой большой выручкой
среднюю выручку на один заказ в каждый день
дни, где возвратов больше 20% заказов

````python
days = [
    {"day": "пн", "orders": 20, "revenue": 40_000, "returns": 2},
    {"day": "вт", "orders": 16, "revenue": 19_200, "returns": 4},
    {"day": "ср", "orders": 25, "revenue": 55_000, "returns": 1},
    {"day": "чт", "orders": 10, "revenue": 12_000, "returns": 3},
    {"day": "пт", "orders": 30, "revenue": 48_000, "returns": 3},
]
all_revenue = 0
max_revenue = 0
avg_revenue_day = {}
day_returns20 = []
for i in days:
    all_revenue += i['revenue']
    if i['revenue'] > max_revenue:
        max_revenue = i['revenue']
        day_max_revenue = i['day']
    avg_revenue_day[i['day']] = round(i['revenue'] / (i['orders'] - i['returns']), 2) #Я тут удали количество возвращенных заказв, решила, так будет лучше. Округлила для симпатичности вывода)
    if i['returns'] / i['orders'] > 0.2:
        day_returns20.append(i['day'])
print(f'''Выручка за всю неделю - {all_revenue}
День с самой большой выручкой - {day_max_revenue}
Средняя выручка на один заказ в каждый день - {avg_revenue_day}
Дни, где возвратов больше 20% заказов - {day_returns20}''')
````

Есть список отзывов на товары. На каждой строке — по одному. Данные слегка битые, и названия товара записаны в разном регистре. id при этом один и тот же для одного и того же товара. Перед подсчётом приведите названия к одному формату, иначе они будут считаться как разные`.

reviews = [
    {"id": 1, "product": "Чехол", "stars": 5},
    {"id": 1, "product": "Чехол", "stars": 3},
    {"id": 1, "product": "Чехол", "stars": 4},
    {"id": 2, "product": "Наушники", "stars": 2},
    {"id": 2, "product": "наушники", "stars": 2},
    {"id": 2, "product": "НАУШНИКИ", "stars": 5},
    {"id": 3, "product": "Планшет", "stars": 5},
    {"id": 4, "product": "Колонка", "stars": 4},
    {"id": 4, "product": "Колонка", "stars": 4},
    {"id": 5, "product": "Кабель", "stars": 1},
]
Нужно найти:

среднюю оценку каждого товара
худший товар по средней оценке среди тех, у кого хотя бы два отзыва
сколько отзывов на 1 или 2 звезды
какую долю всех отзывов они составляют

````python
reviews = [
    {"id": 1, "product": "Чехол", "stars": 5},
    {"id": 1, "product": "Чехол", "stars": 3},
    {"id": 1, "product": "Чехол", "stars": 4},
    {"id": 2, "product": "Наушники", "stars": 2},
    {"id": 2, "product": "наушники", "stars": 2},
    {"id": 2, "product": "НАУШНИКИ", "stars": 5},
    {"id": 3, "product": "Планшет", "stars": 5},
    {"id": 4, "product": "Колонка", "stars": 4},
    {"id": 4, "product": "Колонка", "stars": 4},
    {"id": 5, "product": "Кабель", "stars": 1},
]
bad_stars = 0
set_id = set()
name_stars = {}
for i in reviews:
    i['product'] = i['product'].capitalize()
    name_stars.setdefault(i['product'], []).append(i['stars'])
    set_id.add(i['id'])
    if i['stars'] == 1 or i['stars'] == 2:
        bad_stars += 1
avg_stars_min_two_reviews = {k:sum(v) / len(v) for k, v in name_stars.items() if len(v) > 1} 
min_avg = 5
for k, v in avg_stars_min_two_reviews.items():
    if v < min_avg:
        min_avg = v
        day_min_avg = k

print(f'''Среднюю оценку каждого товара - { {k:sum(v) / len(v) for k, v in name_stars.items()} }
Худший товар по средней оценке среди тех, у кого хотя бы два отзыва - {day_min_avg}
Сколько отзывов на 1 или 2 звезды - {bad_stars}
Какую долю всех отзывов они составляют - {bad_stars / len(reviews)}''')
````
